<span style="font-size:14px">
    <h2>Cliffwalking - Algorithme SARSA</h2>
    <p>Cet algorithme d'<b>apprentissage par renforcement (SARSA)</b> consiste à faire traverser une grille à un agent du début à la fin tout en évitant de tomber d'une falaise.</p>
    <p>Contrairement au Q-Learning qui est "off-policy" le SARSA est "on-policy", ce qui signifie que la politique utilisée pour déterminer les actions de l'agent est la même que celle qui met à jour les valeurs de la Q-table.</p>
    <p>Dans l'état courant (S), l'agent choisit une action (A) et reçoit une récompense (R), il se retrouve dans l'état suivant (S1) et choisit une action (A1) dans S1 (ce qui donne le tuple S, A, R, S1, A1)</p>
    <p style="font-size:14px">L'algorithme SARSA est légèrement différent de celui du Q-Learning dans la mesure où <b>avec SARSA, la valeur Q est mise à jour en tenant compte de l'action A1 effectuée dans l'état S1</b> (alors qu'avec le Q-Learning, l'action ayant la valeur Q la plus élevée dans l'état suivant S1 est utilisée pour mettre à jour la Q-Table). La différence principale entre ces deux algos réside donc dans la manière dont ils mettent à jour la fonction de mise à jour des valeurs de la Q-Table.</p>
    <p>On utilise une <b>Q-Table</b> afin de permettre à l'agent d'atteindre l'objectif situé sur la case [3, 11] de la grille (en évitant de tomber de la falaise, ce qui coûte -100 et fait repartir l'agent de la case départ) en partant de la case [3, 0] en effectuant le moins d'étapes possible (chaque étape coûtant -1) : l'agent applique une stratégie d'exploration décroissante (epsilon-greedy), et la Q-table est mise à jour avec l’action effectivement choisie par l'agent selon cette stratégie.</p>
    <p>On peut ajuster les paramètres (tels que <b>alpha</b>, <b>gamma</b> ou <b>epsilon_decay</b>) afin d'améliorer la convergence et la stabilité de l'apprentissage.</p>
</span>

<h5>Librairies Python</h5>

```python
import numpy as np               # Numerical calculations, matrix management, random number generation, and so on...
import gymnasium as gym          # Provides a pre-designed environment where an agent can learn by interacting with its environment
import matplotlib.pyplot as plt  # 2D graphs generation, data visualization
import seaborn as sns            # Heatmaps (make tables easier to read)
import time                      # Time counting

# Libraries to display all graphs properly on the same picture
import tkinter as tk
from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg
from IPython.display import display
```

<h5>Initialisation de l'environnement</h5>

```python
# is_slippery parameter is False by default
# Default 4x12 map (the only one available)
env = gym.make("CliffWalking-v0", render_mode="rgb_array")

# To view the animated Cliffwalking interface
# env = gym.make("CliffWalking-v0", render_mode="human")
```

<h5>Paramètres de l'algorithme</h5>

```python
alpha = 0.1                 # Learning rate : with a low learning rate, the agent is more stable, less prone to fluctuations (but slower to learn)
gamma = 0.9                 # Discount rate : determines the relative importance of future rewards compared to immediate rewards : should be as close to 1 as possible to avoid neglecting future rewards
epsilon = 1.0               # Epsilon-greedy parameter (epsilon starts to 1 <=> total exploration)
epsilon_min = 0.01          # Minimum epsilon value (minimum exploration probability)
epsilon_decay = 0.00005     # Factor controlling the progressive reduction of epsilon over the course of episodes
num_episodes = 5000         # Number of episodes (iterations)
```

<h5>Stockage des performances de l'algorithme</h5>

```python
rewards_per_episode = np.zeros(num_episodes)    # Table holding the number of rewards per episode
steps_per_episode = np.zeros(num_episodes)      # Table holding the number of actions done per episode
epsilon_values = np.zeros(num_episodes)         # Table holding the epsilon value per episode
start_time = time.time()                        # Training time counter
```

<h5>Initialisation de la Q-Table</h5>

<p style="font-size:14px"><b>Taille de la Q-Table = (Nombre d'états, Nombre d'actions)</b></p>

<p style="font-size:14px">L'environnement CliffWalking-v0 étant une grille 4x12, il y a 4 x 12 = <b>48 états possibles</b>.</p>

<p style="font-size:14px">L'agent peut effectuer <b>4 actions (haut, droite, bas, gauche)</b>.</p>

<p style="font-size:14px">La Q-Table a donc une taille de <b>48 x 4 = 192 valeurs Q.</b></p>

```python
q_table = np.zeros((env.observation_space.n, env.action_space.n))
```

<h5>Stratégie epsilon-greedy</h5>

<p style="font-size:14px"><b>Au fur et à mesure qu'on fait diminuer epsilon, l'agent privilégiera de plus en plus l'exploitation plutôt que l'exploration au fil de son apprentissage.</b></p>
<p style="font-size:14px">En <b>exploration</b>, l'agent choisit aléatoirement (avec une probabilité de epsilon) parmi les 4 actions possibles, explorant de nouvelles possibilités pour apprendre de meilleures façons d'obtenir des récompenses. Cela permet à l'agent de découvrir de nouvelles stratégies et d'améliorer sa prise de décision au fil du temps.</p>
<p style="font-size:14px">En <b>exploitation</b>, l'agent choisit l’action ayant la plus grande valeur Q parmi les 4 valeurs de la Q-Table pour l’état actuel. Cela signifie que l'agent utilise ses connaissances actuelles pour maximiser les récompenses.</p>

```python
def epsilon_greedy_policy(state, epsilon):
    if np.random.rand() < epsilon:
        return env.action_space.sample()     # Exploration
    else:
        return np.argmax(q_table[state, :])  # Exploitation
```

<h5>Entraînement de l'agent avec l'algo SARSA</h5>

<p style="font-size:14px"><b>On entraîne l'agent sur un nombre 'num_episodes' d'itérations.</b></p>
<p style="font-size:14px"><b>Gym 'truncated':</b> indique que l'agent peut effectuer 200 étapes max lors d'un épisode avant que ce dernier ne se termine automatiquement (pour éviter que l'agent reste bloqué indéfiniment).</p>
<p style="font-size:14px"><b>Gym 'terminated':</b> indique que l'agent a atteint l'objectif pour l'épisode courant (case [3, 11]).</p>
<p style="font-size:14px">Les valeurs de la Q-Table (paires état/action) représentent les récompenses attendues pour l'exécution d'une action dans un état spécifique. L'équation de <b>mise à jour des valeurs de la Q-Table</b> est la suivante :</p>

$$
Q(s, a) \leftarrow Q(s, a) + \alpha \left( R(s, a) + \gamma Q(s', a') - Q(s, a) \right)
$$

<p style="font-size:14px">où <b>s</b> représente l'état courant, <b>a</b> l'action choisie par l'agent, <b>s'</b> l'état suivant dans lequel se trouvera l'agent, <b>a'</b> la meilleure action pour s', <b>R</b> la récompense reçue pour avoir pris l'action dans l'état s.</p>

<p style="font-size:14px">Dans l'algo SARSA, la valeur Q est mise à jour en tenant compte de l'action a' effectuée dans l'état s' (alors qu'avec le Q-Learning, l'action ayant la valeur Q la plus élevée dans l'état suivant s' est utilisée pour mettre à jour la Q-Table).<p>

<p style="font-size:14px">En somme, <b>SARSA utilise l'action réellement prise à l'étape suivante</b> (l'agent se base sur ce qu'il a vraiment fait), et le <b>Q-Learning utilise la meilleure action possible à l’étape suivante</b> (l'agent se base sur ce qu'il aurait pu faire de mieux), ce qui rend <b>SARSA plus prudent et réaliste</b>, tandis que le <b>Q-Learning est plus optimiste</b>.</p>

<p style="font-size:14px">A chaque épisode, epsilon est diminué de 'epsilon_decay', ce qui réduit la probabilité d'exploration et augmente celle d'exploitation des connaissances acquises dans la Q-Table. Cela permet à l'agent d'explorer au début, puis de se concentrer sur les actions optimales à mesure qu'il apprend.</p>

```python
for episode in range(num_episodes):
    state = env.reset()[0]
    terminated = False
    truncated = False
    nb_steps = 0            # Number of steps for this episode
    total_reward = 0        # Total reward for this episode
    action = epsilon_greedy_policy(state, epsilon)
    
    while (not terminated and not truncated):
        # The env.step method returns :
        # - The agent's new state
        # - The reward earned for this action
        # - A Boolean indicating whether the episode is over (and metadata is ignored).
        next_state, reward, terminated, truncated, _ = env.step(action)
        next_action = epsilon_greedy_policy(next_state, epsilon)
        
        # Q-Table update
        q_table[state, action] = q_table[state, action] + alpha * (reward + gamma * q_table[next_state, next_action] - q_table[state, action])
        
        state = next_state
        action = next_action
        nb_steps += 1
        total_reward += reward
        # -1 per step unless other reward is triggered
        # -100 for stepping into the cliff

    # Statistics for this episode / Updates
    rewards_per_episode[episode] = total_reward
    steps_per_episode[episode] = nb_steps
    epsilon = max(epsilon_min, epsilon - epsilon_decay)
    epsilon_values[episode] = epsilon

    # Only captures last frame of the last episode
    if episode == (num_episodes - 1): 
        last_frame_img = env.render()

end_time = time.time()
duration = end_time - start_time
env.close()
```

<h5>Affichage des performances de l'algorithme</h5>

```python
print("Mean rewards per episode :", np.mean(rewards_per_episode))
print("Mean number of steps per episode :", np.mean(steps_per_episode))
print("Total training duration :", duration, "secondes")
```

<h2>Affichage des résultats</h2>
<p style="font-size:14px">On regroupe l'ensemble des tableaux et graphes obtenus sur une même image.</p>

<h5>Heatmap de la Q-Table</h5>

<p style="font-size:14px">Permet de visualiser les <b>actions privilégiées par l'agent pour chaque état</b>, et <b>à quel point chaque action choisie est estimée rentable pour un état</b> (les valeurs élevées de la Q-Table sont affichées avec des couleurs chaudes, et les valeurs faibles avec des couleurs froides). Ces actions (directions) privilégiées par l'agent sont représentées par des flèches (les cases proches du vide auront souvent des flèches s’éloignant du bord pour éviter les -100 de pénalité).</p>

<p style="font-size:14px">
La <b>Q-Table Map</b> a donc une taille de 4x12 (correspondant à la taille de la grille représentant l'environnement). Autrement dit, chaque cellule de la Q-Table Map représente l'état associé à une case dans la grille (48 états en tout).
</p>

```python
def plot_q_table_heatmap():

    # Dictionary for mapping actions to arrows
    action_to_arrow = {
        0: "↑",  # Action 0: up
        1: "→",  # Action 1: right
        2: "↓",  # Action 2: down
        3: "←",  # Action 3: left
    }

    q_values_map = np.zeros((4, 12))

    for state in range (env.observation_space.n):
        row = state // 12
        col = state % 12
        q_values_map[row, col] = np.max(q_table[state, :]) # Store the highest Q-value

    fig, ax = plt.subplots(figsize=(12, 4))
    
    # Draw the Q-values heatmap with a colour scale
    # Disable Q-values display as arrows will be displayed instead, for more clarity
    sns.heatmap(q_values_map, annot=False, cmap="coolwarm", cbar=True, square=True, linewidths=0.5)

    # Manually add arrows on the heatmap
    for state in range(env.observation_space.n):
        row = state // 12
        col = state % 12
        best_action = np.argmax(q_table[state, :]) # Find the action with the highest Q-value
        arrow = action_to_arrow[best_action]
        ax.text(col + 0.5, row + 0.5, arrow, ha='center', va='center', fontsize=16, color='black')

    plt.title("Final Q-Table Heatmap")
    plt.xlabel("Actions")
    plt.ylabel("States")
    
    return fig
```

<h5>Graphe représentant l'évolution des récompenses par épisode au fil des itérations</h5>

<p style="font-size:14px">On fait une <b>moyenne glissante</b> pour que le graphe soit lisible : un point représente 100 épisodes.</p>

```python
def plot_rewards():
    rolling_average_rewards_per_episode = np.convolve(rewards_per_episode, np.ones(100)/100, mode='valid')
    
    fig = plt.figure()

    plt.plot(rolling_average_rewards_per_episode)
    plt.title("Rewards per episode (rolling average over 100 episodes)")
    plt.xlabel("Episode")
    plt.ylabel("Rewards")
    plt.grid(True)
    
    return fig
```

<h5>Graphe représentant l'évolution du nombre d'étapes nécessaires pour atteindre la récompense au fil des itérations</h5>

<p style="font-size:14px">On fait une <b>moyenne glissante</b> pour que le graphe soit lisible : un point représente 100 épisodes.</p>

```python
def plot_steps():
    rolling_average_steps_per_episode = np.convolve(steps_per_episode, np.ones(100)/100, mode='valid')

    fig = plt.figure()
    
    plt.plot(rolling_average_steps_per_episode)
    plt.title("Number of steps per episode (rolling average over 100 episodes)")
    plt.xlabel("Episode")
    plt.ylabel("Number of steps")
    plt.grid(True)

    return fig
```

<h5>Graphe représentant l'évolution de la valeur d'epsilon au fur et à mesure des épisodes</h5>

```python
def plot_epsilon():
    fig = plt.figure()

    plt.plot(epsilon_values)
    plt.title("Epsilon evolution through iterations")
    plt.xlabel("Episode")
    plt.ylabel("Epsilon value")
    plt.grid(True)

    return fig
```

<h5>Image représentant la dernière frame de l'apprentissage</h5>

```python
def get_last_frame():
    fig = plt.figure()
    
    plt.imshow(last_frame_img)
    plt.axis("off")
    plt.title("Last Cliffwalking frame")

    return fig
```

<h5>Regroupement des résultats</h5>

```python
# 3x2 grid initialization
fig, axes = plt.subplots(3, 2, figsize=(13, 15))
fig.suptitle("Cliffwalking SARSA Results", fontsize=16, fontweight="bold")

fig1 = plot_q_table_heatmap()
fig2 = plot_rewards()
fig3 = plot_steps()
fig4 = plot_epsilon()
fig5 = get_last_frame()

# Force figures render to avoid issues
for figure in [fig1, fig2, fig3, fig4, fig5]:
    figure.canvas.draw()

# Display graphs on corresponding axes
axes[0, 0].imshow(fig1.canvas.buffer_rgba())
axes[0, 0].set_title("Q-Table Heatmap")

axes[0, 1].imshow(fig2.canvas.buffer_rgba())
axes[0, 1].set_title("Rewards per Episode")

axes[1, 0].imshow(fig3.canvas.buffer_rgba())
axes[1, 0].set_title("Steps per Episode")

axes[1, 1].imshow(fig4.canvas.buffer_rgba())
axes[1, 1].set_title("Epsilon Decay")

axes[2, 0].imshow(fig5.canvas.buffer_rgba())
axes[2, 0].set_title("Final Frame")

# Delete unused space (empty graph)
fig.delaxes(axes[2, 1])

# Delete axes and graduations marks around images
for ax in axes.flatten():  
    ax.axis("off")  

plt.tight_layout()
plt.show()
```
